```python
import numpy as np
from tqdm import tqdm

from milvus_vector_database.src.milvus.milvus_glove import MilvusGlove
from milvus_vector_database.scripts.load_glove_100_angular import load_glove_split
```

# Create milvus client

```python
milvus_glove = MilvusGlove()
```

# Load test data

The train vectors should already be stored in the milvus vector database.

```python
test = load_glove_split('test')
test = test['test']

neighbors = load_glove_split('neighbors')
neighbors = neighbors['neighbors']

print(neighbors[0].keys())
```

# Run search on all test data

```python
targets = []
search_results = []

chunk_size = 100
k = 10
for i in tqdm(range(0, len(test), chunk_size)):
    # Query closest embeddings
    res = milvus_glove.search_vectors(query_vector=test[i:i+chunk_size]['emb'], k=k)
    # Save targets and closest matches
    targets += [n[0] for n in neighbors[i:i+chunk_size]['neighbors_id']]
    search_results += [[entity['id'] for entity in r] for r in res]
```

```python
correct = np.array([targets[i] in search_results[i] for i in range(len(targets))])
np.mean(correct)
```
