# Inspect neighbors data

Meaning keys:
* idx: identifier for sample in the test split.
* neighbors_id: list of 1000 identifiers for the samples in the train split ordered by closest match first.
* neighbors_distance: list of 1000 cosine distances again ordered by closest match first.
* metric: string for the metric for the distance used.
* query_expr: seems to be empty.
* pk_field_name: name of the primary key. I.e. 'idx'.
* vector_field_name: name of the embedding field. I.e. 'emb'.
* top_k: indicates the length of the lists in neighbors_id and neighbors_distance.
